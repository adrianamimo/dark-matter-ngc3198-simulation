```python
import numpy as np
import plotly.graph_objects as go

n = 16000
r_min = 3.086e19
r_max = 1.1e21
a = 9.7384e20
p_0 = 1.049e-22

np.random.seed(60)

b = r_max / 5
radius = np.random.exponential(scale=b, size=n)
radius = np.clip(radius, r_min, r_max)

theta = np.random.uniform(0, 2 * np.pi, n)
phi = np.arccos(1 - 2 * np.random.rand(n))
x = radius * np.sin(phi) * np.cos(theta)
y = radius * np.sin(phi) * np.sin(theta)
z = radius * np.cos(phi)

densities = p_0 / ((radius / a) * (1 + radius / a)**2)

x_kpc = x / 3.086e19
y_kpc = y / 3.086e19
z_kpc = z / 3.086e19

fig = go.Figure(data=go.Scatter3d(
    x=x_kpc,
    y=y_kpc,
    z=z_kpc,
    mode="markers",
    marker=dict(
        color=np.log10(densities),
        colorbar=dict(title="Log₁₀(Density)(kg/m³)"),
        colorscale="Viridis",
        opacity=0.7,
        size=2
    )
))
fig.update_layout(
    title="3D density map of dark matter (NFW profile)",
    font=dict(color="black"),
    paper_bgcolor="white",
    width=800,
    height=700,
    scene=dict(
        xaxis=dict(
            title=dict(
                text="x-axis (kpc)"
            )
        ),
        yaxis=dict(
            title=dict(
                text="y-axis (kpc)"
            )
        ),
        zaxis=dict(
            title=dict(
                text="z-axis (kpc)"
            )
        ),
    ),
)

fig.show()
```
